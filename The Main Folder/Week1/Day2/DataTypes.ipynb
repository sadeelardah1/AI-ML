Sequence Types :

1. List  

An ordered list of items, surrounded by [ ], and Vic:

Add to it  
Delete from it  
Modify its elements  
Arrange it  
Repeat in it  

Example :

```python
#Example :
print("Declaring a list of fruits")
Fruites = ["Apple", "Banana", "Cherry", "Watermelon", "Mango"] 
print(Fruites)
#print(type(Fruites))
print("------------------------------")
#Indexing
print("Indexing the list of fruits")
print(Fruites[0]) #Apple
print(Fruites[1]) #Banana
print(Fruites[2]) #Cherry
print(Fruites[3]) #Watermelon
print(Fruites[4]) #Mango


print("------------------------------")
#Slicing
print("Slicing the list of fruits")
print(Fruites[1:3])   
print(Fruites[:2])    
print(Fruites[2:])    
print(Fruites[:])    
```

List Methods :  
1. .append(x)

    Takes: one element x  
    Returns: None  
    Does: adds x to the end of the list  

2. .insert(i, x)

    Takes: position i + element x  
    Returns: None  
    Does: inserts x at position i  

3. .remove(x)

    Takes: value x  
    Returns: None  
    Does: removes the first element matching x (raises ValueError if not found)  

4. .pop(i)

    Takes: position i (optional, defaults to the last item)  
    Returns: the removed element itself  
    Does: removes and returns the item at position i  

5. .sort()

    Takes: nothing (or optional reverse=True)  
    Returns: None  
    Does: sorts the list in ascending order (permanently)  

6. .reverse()

    Takes: nothing  
    Returns: None  
    Does: reverses the list's order (permanently)  

7. .index(x)

    Takes: value x  
    Returns: the position (int)  
    Does: doesn't change the list — searches for and returns the position of the first occurrence of x  

8. .count(x)

    Takes: value x  
    Returns: number of occurrences (int)  
    Does: doesn't change the list — returns how many times x appears  

```python
Fruites = ["Apple", "Banana", "Cherry", "Watermelon", "Mango"] 
print(Fruites)
Fruites.append("Orange") #Adding an item to the list
Fruites.append("Watermelon") #Adding an item to the list
print(Fruites)
Fruites.insert(2, "Grapes") #Adding an item to the list at a specific index
print(Fruites)
Fruites.remove("Banana") #Removing an item from the list
print(Fruites)
Fruites.pop(4) #Removing an item from the list at a specific index
print(Fruites)
Fruites.sort() #Sorting the list
print(Fruites)
Fruites.reverse() #Reversing the list
print(Fruites)
Fruites.index("Cherry") #Finding the index of an item in the list
print(Fruites)
print(Fruites.count("Watermelon")) #Counting the number of occurrences of an item in the list
```

Loop With List 

```python
Fruites = ["Apple", "Banana", "Cherry", "Watermelon", "Mango"] 
for f in Fruites:
    print(f) #Printing each item in the list
```

2. Tuples
    The list is received exactly in the form, only surrounded by ( ) instead of [ ], and the essential difference is: completely immutable — you never add, delete, or modify it after what you do , use iust index() , count() methods with tuples .  

    Note : 

    General Form :
    Name_of_tuples = (value1 , value2 , value3)  

    Example : 
    

```python
Days = ("Friday", "Saturday", "Sunday","Monday", "Tuesday", "Wednesday", "Thursday")
for d in Days:
    print(d) #Printing each item in the tuple
print("------------------------------")
#Indexing and Slicing in Tuples
print(Days[0]) 
print(Days[-5])
print(Days[2:]) 
print(Days[:3])
```

Tuples Methods  :  
    1. .count(x)  
    2. .index(x)  

Example :  



```python
Days = ("Friday", "Saturday", "Sunday","Monday", "Tuesday", "Wednesday", "Thursday")
print(Days.count("Friday")) #Counting the number of occurrences of an item in the tuple
print(Days.index("Wednesday")) #Finding the index of an item in the tuple
```

Convert btw List & Tuples


```python
my_list = [1, 2, 3]
my_tuple = tuple(my_list)   # تحويل list → tuple
print(my_tuple)              # (1, 2, 3)

back_to_list = list(my_tuple)  # تحويل tuple → list
print(back_to_list)             # [1, 2, 3]
```

Mapping Type :  

1. Dict
    Data stored as key pairs: value, surrounded by { }. Instead of giving the item its location number (index), give it its name (key).  


    Dict Methods   
        1. .keys() : return all keys.  
        2. .values() : return all values .  
        3. .items() : return all pairs (key,value)  

```python
person = {
    "name": "Sadeel",
    "age": 25,
    "city": "Ramallah"
}

print(person)
print(type(person))  # <class 'dict'>

#Access to values :  
print(person["name"])  
print(person["age"])    

#Changing the value of a key in the dictionary
person["age"] = 21
person["city"] = "Jenin"

# To add a new key-value pair to the dictionary, you can simply assign a value to a new key:
person["country"] = "Palestine"
person["GPA"] = 3.8
person["major"] = "Computer Science"

#to delete a key-value pair from the dictionary, you can use the del statement:
del person["major"] # del : delete the key-value pair with the key "major" from the dictionary
print(person.pop("GPA")) # pop : delete the key-value pair with the key "GPA" from the dictionary and return its value. If the key is not found, it raises a KeyError.
print(person)
```

```python
#Methods to get all keys, values, and key-value pairs from the dictionary:
person = {"name": "Sadeel", "age": 21, "city": "jenin"}

print(person.keys())    
print(person.values())  
print(person.items())
```

```python
#looping through the dictionary:
for key, value in person.items():
    print(f"{key}: {value}")
```

Set Type :

1. set 
    A set of values without a specific order and without repetition, surrounded by {} (P.S. without keys, had the difference from dict).  
    - The most important feature inside it is the deletion of repetitions   
    - What is in the index because the set is not ordered-means what you call fruits[0] from the list, because what is in the" first item " is already in a fixed order

```python
# To Create a Set in Python, you can use curly braces {} or the set() constructor. Here's an example of creating a set using both methods:
# Using curly braces
numbers = {1, 2, 2, 3, 2, 3, 4, 5}
print(numbers)  # Output: {1, 2, 3, 4, 5} (duplicates are removed)

#to convert a list to a set, you can use the set() constructor. Here's an example:
# Using the set() constructor
nums = [1, 2, 2, 3, 4, 4, 5]
unique_nums = set(nums)
print(unique_nums)  

# To add an item to a set, you can use the add() method. Here's an example:
fruits = {"apple", "banana", "cherry"}
fruits.add("orange")
print(fruits)  

#methods to remove items from a set:
# Using the remove() method , if the item is not found, it raises a KeyError.
fruits.remove("banana")
print(fruits)  

# Using the discard() method if the item is not found, it does nothing (no error is raised).
fruits.discard("cherry")
print(fruits)  
```

```python
#Set Operations:
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print(a | b)   # Union (اتحاد) → {1, 2, 3, 4, 5, 6}  -- كل القيم من الاثنين
print(a & b)   # Intersection (تقاطع) → {3, 4}  -- القيم المشتركة بس
print(a - b)   # Difference (فرق) → {1, 2}  -- يلي بـ a وما بـ b
```
