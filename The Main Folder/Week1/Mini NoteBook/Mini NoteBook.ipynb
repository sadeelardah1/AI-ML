## **<span style="color:#38BDF8">Sequence Types :</span>**

### **<span style="color:#FACC15">1. List</span>**  

An ordered, **mutable (changeable)** collection of items, surrounded by `[ ]`. You can:

- Add items to it  
- Delete items from it  
- Modify its elements  
- Sort / reorder it  
- Have repeated (duplicate) values in it  

Example :

````python
#Example :
print("Declaring a list of fruits")
Fruites = ["Apple", "Banana", "Cherry", "Watermelon", "Mango"] 
print(Fruites)
print("------------------------------")
#Indexing
print("Indexing the list of fruits")
print(Fruites[0]) #Apple
print(Fruites[1]) #Banana
print(Fruites[2]) #Cherry
print(Fruites[3]) #Watermelon
print(Fruites[4]) #Mango


print("------------------------------")
#Slicing
print("Slicing the list of fruits")
print(Fruites[1:3])   
print(Fruites[:2])    
print(Fruites[2:])    
print(Fruites[:])    
````

#### **<span style="color:#BE185D">List Methods :</span>**  
1. .append(x)

    Takes: one element x  
    Returns: None  
    Does: adds x to the end of the list  

2. .insert(i, x)

    Takes: position i + element x  
    Returns: None  
    Does: inserts x at position i  

3. .remove(x)

    Takes: value x  
    Returns: None  
    Does: removes the first element matching x (raises ValueError if not found)  

4. .pop(i)

    Takes: position i (optional, defaults to the last item)  
    Returns: the removed element itself  
    Does: removes and returns the item at position i  

5. .sort()

    Takes: nothing (or optional reverse=True)  
    Returns: None  
    Does: sorts the list in ascending order (permanently)  

6. .reverse()

    Takes: nothing  
    Returns: None  
    Does: reverses the list's order (permanently)  

7. .index(x)

    Takes: value x  
    Returns: the position (int)  
    Does: doesn't change the list — searches for and returns the position of the first occurrence of x  

8. .count(x)

    Takes: value x  
    Returns: number of occurrences (int)  
    Does: doesn't change the list — returns how many times x appears  

Example :  

````python
Fruites = ["Apple", "Banana", "Cherry", "Watermelon", "Mango"] 
print(Fruites)
Fruites.append("Orange") #Adding an item to the list
Fruites.append("Watermelon") #Adding an item to the list
print(Fruites)
Fruites.insert(2, "Grapes") #Adding an item to the list at a specific index
print(Fruites)
Fruites.remove("Banana") #Removing an item from the list
print(Fruites)
Fruites.pop(4) #Removing an item from the list at a specific index
print(Fruites)
Fruites.sort() #Sorting the list
print(Fruites)
Fruites.reverse() #Reversing the list
print(Fruites)
Fruites.index("Cherry") #Finding the index of an item in the list
print(Fruites)
print(Fruites.count("Watermelon")) #Counting the number of occurrences of an item in the list
````

#### **<span style="color:#BE185D">Looping Through a List</span>**  
You can visit every item in a list using a `for` loop.  
Example :  

````python
Fruites = ["Apple", "Banana", "Cherry", "Watermelon", "Mango"] 
for f in Fruites:
    print(f) #Printing each item in the list
````

### **<span style="color:#FACC15">2. Tuple</span>**  

Just like a list, but surrounded by `( )` instead of `[ ]`. The key difference: a tuple is **immutable** — once it is created, you can never add, remove, or modify its items.  
Because of this, a tuple only supports two methods: `.index()` and `.count()`.  

**General Form :**  
`Name_of_tuple = (value1, value2, value3)`  

Example :  

````python
Days = ("Friday", "Saturday", "Sunday","Monday", "Tuesday", "Wednesday", "Thursday")
for d in Days:
    print(d) #Printing each item in the tuple
print("------------------------------")
#Indexing and Slicing in Tuples
print(Days[0]) 
print(Days[-5])
print(Days[2:]) 
print(Days[:3])
````

#### **<span style="color:#BE185D">Tuple Methods :</span>**  
1. .count(x) — returns how many times x appears  
2. .index(x) — returns the position of the first occurrence of x  

Example :  

````python
Days = ("Friday", "Saturday", "Sunday","Monday", "Tuesday", "Wednesday", "Thursday")
print(Days.count("Friday")) #Counting the number of occurrences of an item in the tuple
print(Days.index("Wednesday")) #Finding the index of an item in the tuple
````

#### **<span style="color:#4ADE80">Bonus : Converting Between List & Tuple</span>**  
Use `tuple()` to convert a list into a tuple, and `list()` to convert a tuple back into a list.  
Example :

````python
my_list = [1, 2, 3]
my_tuple = tuple(my_list)      # convert list -> tuple
print(my_tuple)                # (1, 2, 3)

back_to_list = list(my_tuple)  # convert tuple -> list
print(back_to_list)            # [1, 2, 3]
````

## **<span style="color:#38BDF8">Mapping Type :</span>**

### **<span style="color:#FACC15">1. dict (Dictionary)</span>**  
Data stored as `key : value` pairs, surrounded by `{ }`. Instead of accessing an item by its position (index), you access it by its **name (key)**.  
Example :  

````python
person = {
    "name": "Sadeel",
    "age": 25,
    "city": "Ramallah"
}

print(person)
print(type(person))  # <class 'dict'>

#Access to values :  
print(person["name"])  
print(person["age"])    

#Changing the value of a key in the dictionary
person["age"] = 21
person["city"] = "Jenin"

# To add a new key-value pair, simply assign a value to a new key:
person["country"] = "Palestine"
person["GPA"] = 3.8
person["major"] = "Computer Science"

# del : delete the key-value pair with the key "major"
del person["major"]
# pop : delete the key-value pair with the key "GPA" and return its value (raises KeyError if not found)
print(person.pop("GPA"))
print(person)
````

#### **<span style="color:#BE185D">Dict Methods :</span>**  
1. .keys() — returns all keys  
2. .values() — returns all values  
3. .items() — returns all (key, value) pairs  

Example :  

````python
person = {"name": "Sadeel", "age": 21, "city": "jenin"}

print(person.keys())    
print(person.values())  
print(person.items())
````

#### **<span style="color:#BE185D">Looping Through a Dictionary</span>**  
Use `.items()` to loop over both the key and the value at the same time.  
Example :  

````python
#looping through the dictionary:
for key, value in person.items():
    print(f"{key}: {value}")
````

## **<span style="color:#38BDF8">Set Type :</span>**

### **<span style="color:#FACC15">1. set</span>**  

A collection of values with **no specific order** and **no repetition**, surrounded by `{ }` (note: no keys — this is what makes it different from a dict).  

- Its most important feature is that it automatically **removes duplicates**.  
- Because a set is not ordered, it has no index — you can't access an item like `fruits[0]` the way you would with a list.  

Example :

````python
# To create a set, you can use curly braces {} or the set() constructor.
# Using curly braces
numbers = {1, 2, 2, 3, 2, 3, 4, 5}
print(numbers)  # Output: {1, 2, 3, 4, 5} (duplicates are removed)

# To convert a list to a set, use the set() constructor
nums = [1, 2, 2, 3, 4, 4, 5]
unique_nums = set(nums)
print(unique_nums)  

# To add an item to a set, use the .add() method
fruits = {"apple", "banana", "cherry"}
fruits.add("orange")
print(fruits)  

# .remove(x) : raises a KeyError if the item is not found
fruits.remove("banana")
print(fruits)  

# .discard(x) : does nothing if the item is not found (no error)
fruits.discard("cherry")
print(fruits)  
````

#### **<span style="color:#BE185D">Set Operations</span>**  
- `|` Union — all values from both sets  
- `&` Intersection — only the values common to both sets  
- `-` Difference — values that are in the first set but not in the second  

Example :

````python
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print(a | b)   # Union        -> {1, 2, 3, 4, 5, 6}
print(a & b)   # Intersection -> {3, 4}
print(a - b)   # Difference   -> {1, 2}
````

## **<span style="color:#38BDF8">2. for Loop</span>**  



````python
#for loop to find the sum of even numbers from 1 to 10 and sum of odd numbers from 1 to 10
esum =0 
osum =0
for i in range(1, 11):
    if i % 2 ==0:
        esum+=i
    else:
        osum+=i
        
print(f"Sum Of Even Numbers :{esum}")
print(f"Sum Of Odd Numbers : {osum}")
    
````

## **<span style="color:#38BDF8">3. while Loop</span>**  



````python
# while loop
# Example 1: countdown
count = 10
while count > 0:
    print(count)
    count -= 1
print("Countdown finished!")

# Example 2: password attempts without interactive input
password_attempts = ["1111", "2222", "1234"]
attempt_index = 0
password = ""

while password != "1234" and attempt_index < len(password_attempts):
    password = password_attempts[attempt_index]
    print(f"Trying password: {password}")
    attempt_index += 1

if password == "1234":
    print("Password correct!")
else:
    print("Password not found.")
````

## **<span style="color:#38BDF8">4. Break & Continue</span>**  



````python
#Search the numbers from 1 to 100 and reject the odd numbers and if found a number that is divisible by 4 and >50, break the loop and print the number.
for i in range(1,101):
    if i % 2 != 0 :
        continue 
    
    if i > 50 and i % 4==0:
        print(f"The number that is divisible by 4 and > 50 : {i} \nThe number / 4 : {i/4}")
        break 
````

## **<span style="color:#38BDF8">2. Basic Syntax**</span>
#### Every function starts with the keyword `def`, followed by a name, parentheses `()`, and a colon `:`. The code inside the function must be indented.

**<span style="color:#BE185D">Structure:**</span>  
**<span style="color:#BE185D">def function_name(parameters):**</span>  
**<span style="color:#BE185D">&nbsp;&nbsp;&nbsp;&nbsp;# code to run**</span>  
**<span style="color:#BE185D">&nbsp;&nbsp;&nbsp;&nbsp;return result**</span>

````python
# A simple function that greets a user
def greet(name):
    return f"Hello, {name}! Welcome."

# Calling (using) the function
print(greet("Sadeel"))
print(greet("Ahmad"))
````

## **<span style="color:#38BDF8">3. Parameters vs. Arguments**</span>
#### A **parameter** is the variable name listed inside the parentheses when defining a function. An **argument** is the actual value you pass in when calling it.

**<span style="color:#4ADE80">Note: `name` above is a parameter. `"Sadeel"` is the argument.</span>**

````python
# A function with two parameters
def add_numbers(a, b):
    return a + b

result = add_numbers(5, 3)
print("Sum:", result)
````

## **<span style="color:#38BDF8">4. Default Parameter Values**</span>
#### You can give a parameter a default value. If the caller doesn't provide that argument, the default is used instead.

**<span style="color:#FACC15">This is useful for optional settings, like a discount rate or a greeting message.**</span>

````python
# 'tax_rate' has a default value of 0.10 (10%)
def calculate_total(price, tax_rate=0.10):
    return price + (price * tax_rate)

print(calculate_total(100))          # uses default tax_rate
print(calculate_total(100, 0.05))    # overrides default tax_rate
````

## **<span style="color:#38BDF8">5. The `return` Statement**</span>
#### `return` sends a value back to wherever the function was called. Once `return` runs, the function stops immediately.

**<span style="color:#BE185D">Note: A function with no `return` statement gives back `None` by default.**</span>

````python
# Function WITH return
def square(x):
    return x ** 2

value = square(4)
print("Squared value:", value)

# Function WITHOUT return (returns None)
def print_message(msg):
    print(msg)

output = print_message("Hi there")
print("What print_message returned:", output)
````

## **<span style="color:#38BDF8">6. Docstrings — Documenting Your Function**</span>
#### A docstring is a short description placed right under the `def` line, inside triple quotes `""" """`. It explains what the function does — good practice for professional, readable code.

**<span style="color:#4ADE80">You can view a function's docstring anytime with `help(function_name)`.</span>**

````python
def normalize(value, min_val, max_val):
    """Scale a value to the 0-1 range based on a min and max."""
    return (value - min_val) / (max_val - min_val)

print(normalize(75, 0, 100))  # 0.75

# Viewing the docstring
help(normalize)
````

## **<span style="color:#38BDF8">7. Multiple Return Values**</span>
#### Python lets you return more than one value at once, separated by commas. Behind the scenes, this creates a **tuple**.

**<span style="color:#FACC15">This is very common in data science, e.g. returning both the mean and the max of a dataset.**</span>

````python
def get_stats(numbers):
    """Return the mean, minimum, and maximum as a dictionary.

    Parameters
    ----------
    numbers : list
        A non-empty list of numeric values.

    Returns
    -------
    dict
        A dictionary containing the mean, minimum, and maximum.
    """
    if not numbers:
        raise ValueError("numbers must not be empty")

    return {
        "mean": sum(numbers) / len(numbers),
        "min": min(numbers),
        "max": max(numbers),
    }

stats = get_stats([10, 20, 30, 40, 50])
print("Mean:", stats["mean"])
print("Min:", stats["min"])
print("Max:", stats["max"])
````

## **<span style="color:#38BDF8">8. Keyword Arguments**</span>
#### You can pass arguments by naming the parameter directly. This makes function calls clearer, especially with many parameters, and lets you pass them in any order.


````python
def create_profile(name, age, city="Unknown"):
    return f"{name}, {age} years old, from {city}"

# Using keyword arguments (order doesn't matter here)
print(create_profile(age=25, name="Layla", city="Nablus"))
print(create_profile(name="Omar", age=30))
````

## **<span style="color:#38BDF8">9. Mini Project — Putting It All Together**</span>
#### A small, professional example: a function that analyzes a list of exam scores.

**<span style="color:#4ADE80">Notice how the docstring, default parameter, and multiple return values are all used together.</span>**

````python
def analyze_scores(scores, passing_grade=60):
    """
    Analyze a list of exam scores.
    Returns the average score and the number of students who passed.
    """
    average = sum(scores) / len(scores)
    passed = len([s for s in scores if s >= passing_grade])
    return average, passed

student_scores = [55, 72, 88, 45, 91, 60, 78]
avg, passed_count = analyze_scores(student_scores)

print(f"Average score: {avg:.2f}")
print(f"Students who passed: {passed_count} out of {len(student_scores)}")
````

# **<span style="color:#38BDF8">List Comprehensions & Object-Oriented Programming (OOP) Basics</span>**

</div>

In Python, there are shorter and cleaner ways to write code than the usual `for` loop.

In this notebook we will cover two important topics:

1. **List Comprehensions** — a fast, one-line way to build a new list.
2. **Object-Oriented Programming (OOP) Basics** — how to group data and behavior together using `class`.

## **<span style="color:#38BDF8">1. List Comprehensions</span>**

A **list comprehension** builds a new list in **one readable line**, instead of writing a multi-line `for` loop.

It is considered **idiomatic ("Pythonic") code**, and you will see it constantly in real data-science work.

**General shape:**
```
new_list = [expression for item in iterable]
```

### **<span style="color:#FACC15">1. Basic Syntax</span>**  
We take every item from a list, apply an operation on it, and collect the results in a new list.  
Example :  

````python
#Example :
numbers = [1, 2, 3, 4, 5]
squares = [n ** 2 for n in numbers]  # for each n in numbers, compute n**2

print("Original numbers:", numbers)
print("Squares:", squares)
````

### **<span style="color:#FACC15">2. List Comprehension with a Condition</span>**  
We can add an `if` at the end to keep only the items that match a condition (this is called **filtering**).  
**Shape:** `[expression for item in iterable if condition]`  
Example :  

````python
#Example :
numbers = list(range(1, 11))
evens = [n for n in numbers if n % 2 == 0]  # keep n only if it is even

print("Numbers 1-10:", numbers)
print("Even numbers only:", evens)
````

### **<span style="color:#FACC15">3. List Comprehension with if / else</span>**  
This is different from filtering: here **every** item stays in the list, but its value changes based on a condition.  
**Shape:** `[value_if_true if condition else value_if_false for item in iterable]`  
Example :  

````python
#Example :
numbers = list(range(1, 11))
labels = ["Even" if n % 2 == 0 else "Odd" for n in numbers]

print(labels)
````

### **<span style="color:#FACC15">4. Nested Loops in a List Comprehension</span>**  
A list comprehension can loop over **two (or more) iterables at once**, exactly like nested `for` loops, but in one line.  
Example :  

````python
#Example :
colors = ["Red", "Blue"]
sizes = ["S", "M", "L"]

combinations = [f"{color}-{size}" for color in colors for size in sizes]
print(combinations)
````

#### **<span style="color:#BE185D">Tip : Loop vs. List Comprehension</span>**  
A list comprehension always does the **same job** as a `for` loop that builds a list — it is just shorter and more readable.  

    Traditional loop:
        squares_loop = []
        for n in range(1, 6):
            squares_loop.append(n ** 2)

    List comprehension (same result, one line):
        squares_comp = [n ** 2 for n in range(1, 6)]

Example :  

````python
#Example : comparing both approaches
# Traditional loop
squares_loop = []
for n in range(1, 6):
    squares_loop.append(n ** 2)

# List comprehension (same result, one line)
squares_comp = [n ** 2 for n in range(1, 6)]

print("Loop result:", squares_loop)
print("Comprehension result:", squares_comp)
print("Are they equal?", squares_loop == squares_comp)
````

#### **<span style="color:#4ADE80">Bonus : Dict & Set Comprehensions</span>**  
The same `[ ]` idea also works with `{ }` to build a **dictionary** or a **set** in one line.  
- Dict comprehension → `{key: value for item in iterable}`  
- Set comprehension → `{expression for item in iterable}`  
Example :  

````python
#Example :
students = ["Ali", "Sara", "Omar"]
name_lengths = {name: len(name) for name in students}  # dict comprehension
print("Dict comprehension:", name_lengths)

nums = [1, 2, 2, 3, 4, 4, 5]
unique_squares = {n ** 2 for n in nums}  # set comprehension (duplicates removed)
print("Set comprehension:", unique_squares)
````

## **<span style="color:#38BDF8">2. Object-Oriented Programming (OOP) Basics</span>**

A **class** is a blueprint that bundles **data (attributes)** and **behavior (methods)** together.

- **Attribute** → a variable that belongs to an object (its data).
- **Method** → a function that belongs to an object (what it can do).
- **Object (instance)** → one specific thing created from the class blueprint.

You will not write many classes early on, but AI/ML frameworks (Scikit-learn estimators, PyTorch models) are all built on classes, so understanding the syntax is essential.

### **<span style="color:#FACC15">1. Creating a Class</span>**  
`__init__` is a special method called the **constructor** — it runs automatically when a new object is created, and sets up its attributes.  
`self` refers to **the object itself**, and must be the first parameter of every method.  
Example :  

````python
#Example :
class DataPoint:
    def __init__(self, x, y):
        self.x = x          # attribute
        self.y = y          # attribute

    def distance_from_origin(self):   # method
        return (self.x ** 2 + self.y ** 2) ** 0.5

p1 = DataPoint(3, 4)     # creating an object (instance) of the class
print("Point coordinates:", p1.x, p1.y)
print("Distance from origin:", p1.distance_from_origin())
````

### **<span style="color:#FACC15">2. A Class with Several Methods</span>**  
A class can have as many methods as needed. Each method can read and change the object's attributes through `self`.  
Example :  

````python
#Example :
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print(f"Deposited {amount}. New balance: {self.balance}")

    def withdraw(self, amount):
        if amount > self.balance:
            print("Insufficient funds!")
        else:
            self.balance -= amount
            print(f"Withdrew {amount}. New balance: {self.balance}")

account = BankAccount("Sadeel", 100)
account.deposit(50)
account.withdraw(30)
account.withdraw(1000)   # not enough balance
````

#### **<span style="color:#BE185D">Important : self and __init__</span>**  
    1. __init__(self, ...)
        Runs automatically when an object is created.
        Used to set the object's starting attributes.

    2. self
        Refers to the current object.
        Must be the first parameter of every method (Python passes it automatically).
        self.balance means "this specific object's balance", not a shared value.

#### **<span style="color:#4ADE80">Bonus : Multiple Objects from the Same Class</span>**  
Each object created from a class has its **own independent copy** of the attributes. Changing one object never affects another.  
Example :  

````python
#Example :
p1 = DataPoint(3, 4)
p2 = DataPoint(6, 8)

print("Point 1 distance:", p1.distance_from_origin())
print("Point 2 distance:", p2.distance_from_origin())
print("Are p1 and p2 the same object?", p1 is p2)
````

<h2 style="color:#BE185D; font-weight:bold; margin-top:30px;">1. Why NumPy Exists — The Big Picture</h2>

<h3 style="color:#FACC15; font-weight:bold; background-color:#1f2933; display:inline-block; padding:4px 10px; border-radius:6px;">1.1 The problem with plain Python lists</h3>

A normal Python <code>list</code> is very flexible: it can hold mixed types (numbers, strings, anything), and it can grow or shrink. But that flexibility comes at a cost — Python has to check the type of every single element every time it does something with it. When you have millions of numbers (which is completely normal in AI/ML), this becomes painfully slow.

<h3 style="color:#FACC15; font-weight:bold; background-color:#1f2933; display:inline-block; padding:4px 10px; border-radius:6px;">1.2 The NumPy solution: the ndarray</h3>

NumPy's core object is called the <b>ndarray</b> ("n-dimensional array"). Think of it as a super-charged list with two big differences:

<ul>
<li><b>Homogeneous</b>: every element must be the <i>same data type</i> (e.g. all integers, or all floats). This is what a spreadsheet column looks like too.</li>
<li><b>Fixed operations run in C, not Python</b>: NumPy is actually written in the C programming language under the hood. So when you tell NumPy "multiply every number in this array by 2", it does that in raw, compiled C code — which can be 10x–100x faster than a Python <code>for</code> loop doing the same thing.</li>
</ul>

<div style="border-left:5px solid #4ADE80; background-color:#eafff2; padding:10px 15px; margin:10px 0; border-radius:4px; color:#065f46;">
<b>Tip:</b> Rule of thumb: if you ever find yourself writing a Python <code>for</code> loop to do math over a list of numbers, stop — there is almost always a faster, more readable NumPy way to do the same thing without a loop. This idea is called <b>vectorization</b>, and it is the single most important habit to build this week.
</div>

<h3 style="color:#FACC15; font-weight:bold; background-color:#1f2933; display:inline-block; padding:4px 10px; border-radius:6px;">1.3 Why this matters for your whole internship</h3>

NumPy isn't just "one more library" — it is the foundation everything else sits on:

<ul>
<li><b>Pandas</b> (tomorrow, Day 4) stores its data internally as NumPy arrays.</li>
<li><b>Scikit-learn</b>, <b>TensorFlow</b>, and <b>PyTorch</b> (later phases) all expect data in NumPy-array-like form, and their internal math is built on the same ideas you'll learn today.</li>
</ul>

<h2 style="color:#BE185D; font-weight:bold; margin-top:30px;">2. Creating Arrays</h2>

<div style="border-left:5px solid #FACC15; background-color:#fffbea; padding:10px 15px; margin:10px 0; border-radius:4px; color:#78350f;">
<b>Goal:</b> Understand every common way to make a NumPy array, and know which one to reach for in which situation.
</div>

There isn't just one way to create an array — different situations call for different creation functions. Let's go through each one slowly.


<h3 style="color:#FACC15; font-weight:bold; background-color:#1f2933; display:inline-block; padding:4px 10px; border-radius:6px;">2.1 From a Python list — np.array()</h3>

The most direct way: you already have a Python list of numbers, and you convert it into an ndarray.


````python
a = np.array([1, 2, 3, 4])
print(a)
print(type(a))
````

<div style="border-left:5px solid #38BDF8; background-color:#eaf6ff; padding:10px 15px; margin:10px 0; border-radius:4px; color:#0c4a6e;">
<b>Note:</b> You can also pass a list-of-lists to create a 2D array (rows and columns) — we will use this a lot below.
</div>

````python
m = np.array([[1, 2, 3],
              [4, 5, 6]])
print(m)
````

<h3 style="color:#FACC15; font-weight:bold; background-color:#1f2933; display:inline-block; padding:4px 10px; border-radius:6px;">2.2 Arrays filled with zeros or ones — np.zeros() / np.ones()</h3>

Very often in ML you need a "blank" array to start with — for example, to store results you will fill in later, or to initialize weights. <code>np.zeros()</code> and <code>np.ones()</code> create arrays pre-filled with 0 or 1.

The argument you pass is the <b>shape</b> — for a 2D array, this is a tuple <code>(rows, columns)</code>.


````python
z = np.zeros((2, 3))
print("zeros:\n", z)

o = np.ones(5)
print("ones:\n", o)
````

<h3 style="color:#FACC15; font-weight:bold; background-color:#1f2933; display:inline-block; padding:4px 10px; border-radius:6px;">2.3 A range of numbers — np.arange()</h3>

This is the NumPy version of Python's built-in <code>range()</code>, but it returns an actual array (and it also accepts decimal steps, which plain <code>range()</code> cannot do).

Syntax: <code>np.arange(start, stop, step)</code> — note that <code>stop</code> is <b>not included</b> (this is called a "half-open interval", and it matches how Python's <code>range()</code> behaves too).


````python
r = np.arange(0, 10, 2)
print(r)
````

<h3 style="color:#FACC15; font-weight:bold; background-color:#1f2933; display:inline-block; padding:4px 10px; border-radius:6px;">2.4 Evenly spaced values — np.linspace()</h3>

<code>np.linspace(start, stop, n)</code> is different from <code>arange</code>: instead of choosing a *step size*, you choose *how many numbers you want*, and NumPy spaces them evenly between start and stop — and here, <code>stop</code> <b>IS included</b>.


````python
l = np.linspace(0, 1, 5) 
print(l)
````

<div style="border-left:5px solid #BE185D; background-color:#fff0f6; padding:10px 15px; margin:10px 0; border-radius:4px; color:#7a0d3d;">
<b>Important:</b> <code>arange</code> vs <code>linspace</code> is a very common point of confusion. Remember it this way: <b>arange</b> = you control the <i>step</i>; <b>linspace</b> = you control the <i>count</i> of numbers.

<h3 style="color:#FACC15; font-weight:bold; background-color:#1f2933; display:inline-block; padding:4px 10px; border-radius:6px;">2.5 Random arrays — np.random.rand()</h3>

Useful for testing code, simulations, or initializing values before training a model. <code>np.random.rand(shape)</code> generates random decimal numbers between 0 (inclusive) and 1 (exclusive).


````python
rnd = np.random.rand(3, 3)
print(rnd)
````

<h2 style="color:#BE185D; font-weight:bold; margin-top:30px;">3. Array Attributes, Indexing &amp; Slicing</h2>

<div style="border-left:5px solid #FACC15; background-color:#fffbea; padding:10px 15px; margin:10px 0; border-radius:4px; color:#78350f;">
<b>Goal:</b> Be able to read the shape/dtype of any array, and pull out exactly the rows, columns, or single values you need.
</div>

<h3 style="color:#FACC15; font-weight:bold; background-color:#1f2933; display:inline-block; padding:4px 10px; border-radius:6px;">3.1 Attributes — asking the array about itself</h3>

Every NumPy array carries some built-in information about itself, called <b>attributes</b>. You don't call these with parentheses (no <code>()</code>) — you just read them like a property.


````python
m = np.array([[1, 2, 3],
              [4, 5, 6]])

print("shape:", m.shape) 
print("ndim :", m.ndim) 
print("dtype:", m.dtype) 
print("size :", m.size) 
````

<div style="border-left:5px solid #38BDF8; background-color:#eaf6ff; padding:10px 15px; margin:10px 0; border-radius:4px; color:#0c4a6e;">
<b>Note:</b> <b>shape</b> = the &quot;dimensions&quot; of the array, written as (rows, columns) for 2D. <b>dtype</b> = what kind of number is stored (integer, float, etc.) — remember, ALL elements in one array must share the same dtype. <b>ndim</b> = how many dimensions (1D, 2D, 3D...). <b>size</b> = the total count of numbers in the array.
</div>

<h3 style="color:#FACC15; font-weight:bold; background-color:#1f2933; display:inline-block; padding:4px 10px; border-radius:6px;">3.2 Indexing — getting ONE specific element</h3>

For a 1D array, indexing works exactly like a Python list: <code>a[0]</code> is the first element.

For a 2D array, you use a <b>comma</b> to separate row and column: <code>array[row, column]</code>. This is different from a nested list, where you would need <code>list[row][column]</code> — NumPy lets you do it in a single set of brackets.


````python
m = np.array([[1, 2, 3],
              [4, 5, 6]])

print(m[0, 2]) 
print(m[1, 0])
````

<h3 style="color:#FACC15; font-weight:bold; background-color:#1f2933; display:inline-block; padding:4px 10px; border-radius:6px;">3.3 Slicing — getting a RANGE of elements</h3>

A slice uses a colon <code>:</code> to mean "everything" or "a range" along that axis. This is one of the most useful things you will do with NumPy every single day.


````python
m = np.array([[1, 2, 3],
              [4, 5, 6]])

print("all rows, column 1:      ", m[:, 1])  
print("row 1, all columns:      ", m[1, :])
print("all rows, first 2 cols:  \n", m[:, 0:2])
````

<div style="border-left:5px solid #38BDF8; background-color:#eaf6ff; padding:10px 15px; margin:10px 0; border-radius:4px; color:#0c4a6e;">
<b>Note:</b> Read <code>m[:, 1]</code> out loud as: &quot;give me : (all) rows, and column 1&quot;. Read <code>m[1, :]</code> as: &quot;give me row 1, and : (all) columns&quot;. Once you can read slices out loud like this, they stop being confusing.
</div>

<h3 style="color:#FACC15; font-weight:bold; background-color:#1f2933; display:inline-block; padding:4px 10px; border-radius:6px;">3.4 Boolean Masking — filtering by a CONDITION</h3>

Instead of selecting by position (index 0, index 1...), you can select by a <i>condition</i> — e.g. "give me every value greater than 2". This works in two steps:
<ol>
<li>Write the condition — this creates a new array of <code>True</code>/<code>False</code> values, one for each element (this is called a <b>boolean mask</b>).</li>
<li>Put that mask inside square brackets on the original array — NumPy keeps only the elements where the mask is <code>True</code>.</li>
</ol>


````python
a = np.array([1, 2, 3, 4, 5, 6])

mask = a > 3
print("the mask itself:", mask)   

filtered = a[mask]
print("filtered values:", filtered) 


print(a[a > 3])
````

<div style="border-left:5px solid #4ADE80; background-color:#eafff2; padding:10px 15px; margin:10px 0; border-radius:4px; color:#065f46;">
<b>Tip:</b> Boolean masking is EXACTLY the same idea as Pandas filtering (which you'll learn tomorrow, Day 4) — e.g. <code>df[df[&quot;age&quot;] &gt; 30]</code>. If you understand masking here, Day 4 filtering will feel completely familiar.

<h2 style="color:#BE185D; font-weight:bold; margin-top:30px;">4. Vectorized Operations</h2>

<div style="border-left:5px solid #FACC15; background-color:#fffbea; padding:10px 15px; margin:10px 0; border-radius:4px; color:#78350f;">
<b>Goal:</b> Stop writing loops for math on arrays — let NumPy apply the operation to every element at once.
</div>

<h3 style="color:#FACC15; font-weight:bold; background-color:#1f2933; display:inline-block; padding:4px 10px; border-radius:6px;">4.1 What &quot;vectorized&quot; means</h3>

A vectorized operation applies to an entire array at once, with no explicit <code>for</code> loop written by you. NumPy handles the looping internally, in fast C code.


````python
a = np.array([1, 2, 3, 4])

print(a * 2)        
print(a + a)       
print(a ** 2)     
````

<h3 style="color:#FACC15; font-weight:bold; background-color:#1f2933; display:inline-block; padding:4px 10px; border-radius:6px;">4.2 Compare: the SLOW loop way vs the FAST vectorized way</h3>

This side-by-side comparison is the whole point of NumPy — same result, very different code (and very different speed on large data).


````python
numbers = [1, 2, 3, 4, 5]


doubled_slow = []
for n in numbers:
    doubled_slow.append(n * 2)
print("loop result:      ", doubled_slow)

arr = np.array(numbers)
doubled_fast = arr * 2
print("vectorized result:", doubled_fast)
````

<div style="border-left:5px solid #BE185D; background-color:#fff0f6; padding:10px 15px; margin:10px 0; border-radius:4px; color:#7a0d3d;">
<b>Important:</b> On a small list of 5 numbers you won't notice a speed difference — but on a dataset of 1,000,000 rows (completely normal in real ML work), the vectorized version can be dozens of times faster, and it's also far more readable.
</div>

<h3 style="color:#FACC15; font-weight:bold; background-color:#1f2933; display:inline-block; padding:4px 10px; border-radius:6px;">4.3 Built-in aggregation methods</h3>

NumPy arrays come with handy built-in methods to summarize all their data at once.


````python
a = np.array([1, 2, 3, 4])

print("mean:", a.mean())  
print("sum :", a.sum())  
print("min :", a.min())  
print("max :", a.max())   
print("std :", a.std())  
````

<h2 style="color:#BE185D; font-weight:bold; margin-top:30px;">5. Broadcasting</h2>

<div style="border-left:5px solid #FACC15; background-color:#fffbea; padding:10px 15px; margin:10px 0; border-radius:4px; color:#78350f;">
<b>Goal:</b> Understand the rule that lets NumPy combine arrays of DIFFERENT shapes — and understand why this is exactly what causes most beginner shape-mismatch errors.
</div>

<h3 style="color:#FACC15; font-weight:bold; background-color:#1f2933; display:inline-block; padding:4px 10px; border-radius:6px;">5.1 The simplest case: array + single number</h3>

You've actually already seen broadcasting without naming it — when we did <code>a * 2</code> above. NumPy "stretched" the single number <code>2</code> across every element of the array. That stretching-to-fit behavior is called <b>broadcasting</b>.

<h3 style="color:#FACC15; font-weight:bold; background-color:#1f2933; display:inline-block; padding:4px 10px; border-radius:6px;">5.2 The more interesting case: 2D array + 1D array</h3>

NumPy can also add a shorter array to a longer/bigger one, IF the shapes are "compatible". The clearest example: adding one row to every row of a 2D array.


````python
matrix = np.array([[1, 2, 3],
                    [4, 5, 6]])  

row = np.array([10, 20, 30])     

result = matrix + row
print(result)
````

<div style="border-left:5px solid #38BDF8; background-color:#eaf6ff; padding:10px 15px; margin:10px 0; border-radius:4px; color:#0c4a6e;">
<b>Note:</b> Picture it as NumPy invisibly &quot;copying&quot; the row <code>[10, 20, 30]</code> to match the matrix's 2 rows, WITHOUT actually using extra memory for real copies — that's why it's efficient, not just convenient.
</div>

<h3 style="color:#FACC15; font-weight:bold; background-color:#1f2933; display:inline-block; padding:4px 10px; border-radius:6px;">5.3 The rule for when broadcasting is allowed</h3>

Two shapes are compatible for broadcasting if, comparing their dimensions from the <b>right</b> side:
<ul>
<li>the dimensions are equal, OR</li>
<li>one of the dimensions is 1 (or missing)</li>
</ul>

Our example above: matrix shape <code>(2, 3)</code> vs row shape <code>(3,)</code> → compare from the right: 3 vs 3 match, and the matrix simply has an "extra" dimension (2) that the row doesn't need. So it works.


````python

try:
    bad = np.array([[1, 2, 3], [4, 5, 6]]) + np.array([1, 2]) 
except ValueError as e:
    print("ValueError:", e)
````

<div style="border-left:5px solid #BE185D; background-color:#fff0f6; padding:10px 15px; margin:10px 0; border-radius:4px; color:#7a0d3d;">
<b>Important:</b> This &quot;operands could not be broadcast together&quot; error is one of the most common errors you will see in your entire AI/ML career. Whenever you see it: STOP, print both arrays' <code>.shape</code>, and compare them from the right-hand side using the rule above.
</div>

## **<span style="color:#38BDF8">0. What Is Pandas?**</span>
#### Pandas is the main Python library for working with tabular (spreadsheet-like) data — think of it as Excel, but inside Python and much more powerful.

**<span style="color:#FACC15">Its two core objects:**</span>  
**<span style="color:#FACC15">- Series : a single labeled column of data**</span>  
**<span style="color:#FACC15">- DataFrame : a full table — rows and named columns, exactly like a CSV, Excel sheet, or SQL query result**</span>

````python
# Always start by importing pandas with the alias 'pd'
import pandas as pd
import numpy as np
````

## **<span style="color:#38BDF8">1. Series vs. DataFrame**</span>
#### A Series is one column with labels (called an index). A DataFrame is a collection of Series sharing the same index — basically a table.

````python
# A Series: one column of data with an index
ages = pd.Series([21, 25, 19, 30], name="age")
print(ages)
print(type(ages))
````

````python
# A DataFrame: a full table made of several columns (Series)
students = pd.DataFrame({
    "name": ["Sara", "Omar", "Layla", "Ahmad"],
    "age": [21, 25, 19, 30],
    "city": ["Nablus", "Ramallah", "Jenin", "Hebron"]
})
print(students)
print(type(students))
````

## **<span style="color:#38BDF8">2. Loading and Inspecting Data**</span>
#### The first step with any dataset is always to load it, then look at it before doing anything else: its shape, its columns, its data types, and a sample of rows.

**<span style="color:#BE185D">Key functions:**</span>  
**<span style="color:#BE185D">- pd.read_csv("file.csv") : loads a CSV file into a DataFrame**</span>  
**<span style="color:#BE185D">- .head() : shows the first 5 rows**</span>  
**<span style="color:#BE185D">- .shape : returns (rows, columns)**</span>  
**<span style="color:#BE185D">- .info() : shows column names, non-null counts, and data types**</span>  
**<span style="color:#BE185D">- .describe() : shows summary statistics for numeric columns**</span>

````python
# We'll simulate a real dataset here instead of reading a CSV file,
# so this notebook runs on its own without needing an external file.
# In real life you would write: df = pd.read_csv("data.csv")

np.random.seed(42)  # fixes randomness so results are repeatable

df = pd.DataFrame({
    "name": ["Sara", "Omar", "Layla", "Ahmad", "Nour", "Khaled", "Rana", "Yousef"],
    "age": [21, 25, 19, np.nan, 30, 22, 28, 24],
    "city": ["Nablus", "Ramallah", "Jenin", "Hebron", "Nablus", "Ramallah", "Jenin", "Nablus"],
    "income": [1200, 1800, 900, 1500, 2000, 1100, np.nan, 1700]
})

df.head()  # first 5 rows
````

````python
print("Shape (rows, columns):", df.shape)
print("------------------------------")
df.info()   # column names, non-null counts, dtypes
````

````python
df.describe()   # summary statistics for numeric columns only
````

## **<span style="color:#38BDF8">3. Selecting and Filtering**</span>
#### You can select columns by name, and select rows by a condition (this is called boolean filtering — the Pandas equivalent of NumPy's boolean masking).

**<span style="color:#FACC15">- .loc[ ] selects by label (column/row name)**</span>  
**<span style="color:#FACC15">- .iloc[ ] selects by integer position**</span>

````python
# Selecting a single column returns a Series
print(df["age"])
````

````python
# Selecting multiple columns returns a DataFrame (note the double brackets)
df[["name", "income"]]
````

````python
# Filtering rows where age is greater than 22
df[df["age"] > 22]
````

````python
# .loc : select the income column, but only for rows where age > 22
df.loc[df["age"] > 22, "income"]
````

## **<span style="color:#38BDF8">4. Cleaning Data**</span>
#### Real data is messy: missing values, duplicates, and wrong types are the norm. Cleaning is often the largest part of any data project, and doing it correctly is what everything downstream depends on.

**<span style="color:#BE185D">- .isnull().sum() : counts missing values per column**</span>  
**<span style="color:#BE185D">- .fillna(value) : fills missing values with something (e.g. the mean)**</span>  
**<span style="color:#BE185D">- .dropna() : removes rows that contain missing values**</span>  
**<span style="color:#BE185D">- .drop_duplicates() : removes duplicate rows**</span>

````python
# Step 1: Count missing values per column
df.isnull().sum()
````

````python
# Step 2: Fill missing 'age' values with the average age
df["age"] = df["age"].fillna(df["age"].mean())

# Step 3: Fill missing 'income' values with the average income
df["income"] = df["income"].fillna(df["income"].mean())

print("Missing values after cleaning:")
df.isnull().sum()
````

````python
# Removing duplicate rows (none exist here, but this is how you would do it)
df = df.drop_duplicates()
df
````

**<span style="color:#4ADE80">Note: We filled the missing values instead of dropping them (df.dropna()), because dropping rows means losing real data. Filling with the mean is a common, simple strategy for numeric columns.</span>**

## **<span style="color:#38BDF8">5. Grouping and Aggregation (groupby)**</span>
#### groupby splits the data into groups based on a column's values, applies an aggregation (mean, sum, count...) to each group, then combines the results. This is the single most useful Pandas operation for finding patterns in data — known as **"split → apply → combine"**.

````python
# Average income per city
df.groupby("city")["income"].mean()
````

````python
# Multiple aggregations at once: average income AND maximum age, per city
df.groupby("city").agg({"income": "mean", "age": "max"})
````

## **<span style="color:#38BDF8">6. Mini Project — Putting It All Together**</span>
#### A small, professional example that follows the same load → inspect → clean → filter → group pipeline used with any real dataset.

````python
# Step 1: Report shape, columns, and dtypes
print("Shape:", df.shape)
print("Columns:", list(df.columns))
print("Dtypes:\n", df.dtypes)
````

````python
# Step 2: Filter to a meaningful subset — people earning above the average income
average_income = df["income"].mean()
high_earners = df[df["income"] > average_income]
print(f"Average income: {average_income:.2f}")
high_earners
````

````python
# Step 3: Aggregate statistic per category, interpreted below
city_summary = df.groupby("city").agg({"income": "mean", "age": "mean"})
city_summary
````

**<span style="color:#FACC15">Interpretation: Nablus has the highest average income among the three cities in this sample, while Jenin has the lowest. This kind of grouped summary is exactly what groupby is used for — spotting patterns across categories at a glance.</span>**

<h2 style="color:#BE185D; font-weight:bold; margin-top:30px;">1. Why Visualization Matters — The Big Picture</h2>

<h3 style="color:#FACC15; font-weight:bold; background-color:#1f2933; display:inline-block; padding:4px 10px; border-radius:6px;">1.1 Numbers hide patterns, charts reveal them</h3>

Imagine looking at a table of 10,000 rows of income data. Just by staring at the numbers, could you tell whether the data is evenly spread, clustered around one value, or has a few extreme outliers? Almost certainly not — but a single histogram would show you the answer <i>instantly</i>. This is the core reason visualization exists: <b>the human eye is extremely good at spotting patterns in shapes, but very bad at spotting patterns in raw numbers.</b>

<h3 style="color:#FACC15; font-weight:bold; background-color:#1f2933; display:inline-block; padding:4px 10px; border-radius:6px;">1.2 Two jobs of a chart: exploration and communication</h3>

Visualization actually does two different jobs, and it helps to keep them separate in your mind:

<ul>
<li><b>Exploration</b> — charts <i>you</i> make for yourself, while you are still investigating the data, to understand trends, clusters, outliers, and relationships between variables.</li>

<h2 style="color:#BE185D; font-weight:bold; margin-top:30px;">3. Line Plots — plt.plot()</h2>

<h3 style="color:#FACC15; font-weight:bold; background-color:#1f2933; display:inline-block; padding:4px 10px; border-radius:6px;">3.1 What a line plot is for</h3>

A <b>line plot</b> connects data points with a line, in the order they are given. It is the natural choice whenever your x-axis represents something <b>continuous and ordered</b> — most commonly <b>time</b>. Use it to show a trend or a change over time.


````python
days = np.arange(1, 8)                       # 1, 2, 3, ..., 7
temperature = [22, 23, 25, 24, 26, 28, 27] 

plt.plot(days, temperature)
plt.xlabel("Day of the Week")
plt.ylabel("Temperature (°C)")
plt.title("Temperature Over One Week")
plt.show()
````

<h3 style="color:#FACC15; font-weight:bold; background-color:#1f2933; display:inline-block; padding:4px 10px; border-radius:6px;">3.2 Plotting more than one line</h3>

You can call <code>plt.plot()</code> multiple times before <code>plt.show()</code> to draw several lines on the same chart — very useful for comparing two related trends.


````python
nablus_temp = [22, 23, 25, 24, 26, 28, 27]
ramallah_temp = [20, 21, 23, 22, 24, 25, 24]

plt.plot(days, nablus_temp, label="Nablus")
plt.plot(days, ramallah_temp, label="Ramallah")
plt.xlabel("Day of the Week")
plt.ylabel("Temperature (°C)")
plt.title("Temperature Comparison: Two Cities")
plt.legend()          # shows which line is which, using the "label" text above
plt.show()
````

<b>Note:</b> Without <code>plt.legend()</code>, the labels you gave with <code>label="..."</code> are stored but never actually shown on the chart — you must always call <code>plt.legend()</code> to make them appear.<br><code>plt.legend()</code> : It is a box that explains the meaning of each line within the drawing.

<h2 style="color:#BE185D; font-weight:bold; margin-top:30px;">4. Scatter Plots — plt.scatter()</h2>

<h3 style="color:#FACC15; font-weight:bold; background-color:#1f2933; display:inline-block; padding:4px 10px; border-radius:6px;">4.1 What a scatter plot is for</h3>

A <b>scatter plot</b> draws each data point as an individual dot, with <i>no</i> line connecting them. It is the best choice for showing the <b>relationship between two numeric variables</b> — for example: does higher study time relate to a higher exam score?


````python
import numpy as np
import matplotlib.pyplot as plt

study_hours = np.array([1, 2, 3, 4, 5, 6, 7, 8])
exam_scores = np.array([50, 55, 65, 68, 75, 80, 88, 94])

plt.figure(figsize=(9, 5))

plt.scatter(study_hours, exam_scores)

plt.title("Study Hours vs Exam Scores")
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")

plt.grid(True)
plt.show()
````

````python
np.random.seed(42)   # fixed seed so this example is reproducible every time

study_hours = np.random.uniform(1, 10, 40)                       # 40 random values between 1 and 10
exam_score = study_hours * 6 + np.random.normal(0, 8, 40) + 30   # a noisy upward relationship

plt.scatter(study_hours, exam_score)
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")
plt.title("Study Hours vs. Exam Score")
plt.show()
````

<h3 style="color:#FACC15; font-weight:bold; background-color:#1f2933; display:inline-block; padding:4px 10px; border-radius:6px;">4.2 Types Of Relationships In Scatter Plot</h3>  

- Positive Relationship : As x increases, y increases
- Negative Relationship  : As x increases, y decreases
- No Clear Relationship : Sometimes there is no clear pattern

<div style="border-left:5px solid #4ADE80; background-color:#eafff2; padding:10px 15px; margin:10px 0; border-radius:4px; color:#065f46;">
<b>Tip:</b> Looking at the scatter plot above, you can visually judge whether there is a positive relationship (points trending upward), a negative relationship (trending downward), or no clear relationship (a random cloud) — this is exactly the kind of pattern a table of numbers would hide.
</div>



<h3 style="color:#FACC15; font-weight:bold; background-color:#1f2933; display:inline-block; padding:4px 10px; border-radius:6px;">4.3 Customizing scatter points</h3>

You can control the size (<code>s</code>), color (<code>c</code>), and transparency (<code>alpha</code>) of the dots — very useful when points overlap.


````python
plt.scatter(study_hours, exam_score, s=80, c="#38BDF8", alpha=0.7, edgecolors="black")
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")
plt.title("Study Hours vs. Exam Score (styled)")
plt.show()
````

<h2 style="color:#BE185D; font-weight:bold; margin-top:30px;">5. Bar Charts — plt.bar()</h2>

<h3 style="color:#FACC15; font-weight:bold; background-color:#1f2933; display:inline-block; padding:4px 10px; border-radius:6px;">5.1 What a bar chart is for</h3>

A <b>bar chart</b> uses the <i>height</i> of a bar to compare a value across separate <b>categories</b> (things that are not numerically continuous, like names, countries, or product types).


````python
tracks = ["AI & ML", "Web Dev", "Cybersecurity", "Data Analysis"]
num_interns = [18, 25, 12, 15]

plt.bar(tracks, num_interns, color="#FACC15")
plt.xlabel("Track")
plt.ylabel("Number of Interns")
plt.title("Interns per Track")
plt.show()
````

<h3 style="color:#FACC15; font-weight:bold; background-color:#1f2933; display:inline-block; padding:4px 10px; border-radius:6px;">5.2 Horizontal bars — plt.barh()</h3>

When category names are long, a horizontal bar chart (<code>plt.barh()</code>) is often easier to read, because the text does not get squeezed or rotated.


````python
plt.barh(tracks, num_interns, color="#4ADE80")
plt.xlabel("Number of Interns")
plt.title("Interns per Track (horizontal)")
plt.show()
````

<b>Important:</b> Never use a line plot for categorical data (like the tracks above) — connecting categories with a line falsely suggests there is an <i>order</i ​ or a <i>trend</i> between them, which does not make sense here. Bar charts are the correct tool for categories.

<h2 style="color:#BE185D; font-weight:bold; margin-top:30px;">6. Histograms — plt.hist()</h2>

<h3 style="color:#FACC15; font-weight:bold; background-color:#1f2933; display:inline-block; padding:4px 10px; border-radius:6px;">6.1 What a histogram is for</h3>

A <b>histogram</b> shows the <b>distribution</b> of a single numeric variable — meaning: how the values are spread out. It works by splitting the data's range into equal-width "bins" and counting how many values fall into each bin.

<div style="border-left:5px solid #38BDF8; background-color:#eaf6ff; padding:10px 15px; margin:10px 0; border-radius:4px; color:#0c4a6e;">
<b>Note:</b> A histogram looks similar to a bar chart, but it is conceptually different: a bar chart compares <i>separate categories</i>, while a histogram shows how <i>one continuous variable</i> is distributed across ranges of values.
</div>


````python
np.random.seed(42)
ages = np.random.normal(loc=25, scale=4, size=300)   # 300 simulated ages, centered around 25

plt.hist(ages, bins=15, color="#BE185D", edgecolor="white") #bins :يعني تقسيم نطاق الدرجات إلى خمس مجموعات
plt.xlabel("Age")
plt.ylabel("Number of Interns")
plt.title("Distribution of Intern Ages")
plt.show()
````

<h3 style="color:#FACC15; font-weight:bold; background-color:#1f2933; display:inline-block; padding:4px 10px; border-radius:6px;">6.2 Choosing the number of bins</h3>

The <code>bins</code> parameter controls how many bars the histogram is split into. Too few bins hide detail; too many bins make the chart look noisy. There is no single "correct" number — try a few values and see what tells the clearest story.


````python
fig, axes = plt.subplots(1, 3, figsize=(14, 4))

axes[0].hist(ages, bins=5, color="#38BDF8")
axes[0].set_title("bins=5 (too few — hides detail)")

axes[1].hist(ages, bins=15, color="#FACC15")
axes[1].set_title("bins=15 (a reasonable choice)")

axes[2].hist(ages, bins=50, color="#4ADE80")
axes[2].set_title("bins=50 (too many — looks noisy)")

plt.tight_layout()
plt.show()
````

<div style="border-left:5px solid #4ADE80; background-color:#eafff2; padding:10px 15px; margin:10px 0; border-radius:4px; color:#065f46;">
<b>Tip:</b> A common, reasonable starting point is <code>bins=10</code> to <code>bins=20</code> for a medium-sized dataset — then adjust by eye.
</div>

<h2 style="color:#BE185D; font-weight:bold; margin-top:30px;">8. Subplots — Multiple Charts in One Figure</h2>

<h3 style="color:#FACC15; font-weight:bold; background-color:#1f2933; display:inline-block; padding:4px 10px; border-radius:6px;">8.1 Why subplots?</h3>

Often you want to place several charts <b>side by side</b>, so a reader can compare them at a glance, instead of scrolling through separate cells. <code>plt.subplots(rows, cols)</code> creates a grid of individual chart areas, called <b>axes</b>, that you then draw into one at a time.

<h3 style="color:#FACC15; font-weight:bold; background-color:#1f2933; display:inline-block; padding:4px 10px; border-radius:6px;">8.2 A 1x2 grid, step by step</h3>


````python
fig, axes = plt.subplots(1, 2, figsize=(11, 4))   # 1 row, 2 columns -> axes is a list of 2 subplot areas

# axes[0] -- the left subplot
axes[0].hist(ages, bins=15, color="#38BDF8")
axes[0].set_title("Age Distribution")
axes[0].set_xlabel("Age")
axes[0].set_ylabel("Count")

# axes[1] -- the right subplot
axes[1].scatter(study_hours, exam_score, color="#BE185D", alpha=0.7)
axes[1].set_title("Study Hours vs. Exam Score")
axes[1].set_xlabel("Study Hours")
axes[1].set_ylabel("Exam Score")

plt.tight_layout()   # automatically prevents titles/labels from overlapping between subplots
plt.show()
````

<div style="border-left:5px solid #BE185D; background-color:#fff0f6; padding:10px 15px; margin:10px 0; border-radius:4px; color:#7a0d3d;">
<b>Important:</b> Notice the subtle syntax difference once you are inside a subplot: use <code>axes[i].set_title()</code>, <code>axes[i].set_xlabel()</code>, <code>axes[i].set_ylabel()</code> — <b>not</b> the plain <code>plt.title()</code> style used for a single chart. Forgetting the <code>set_</code> prefix, or forgetting to target <code>axes[i]</code>, is one of the most common subplot mistakes.
</div>

<h3 style="color:#FACC15; font-weight:bold; background-color:#1f2933; display:inline-block; padding:4px 10px; border-radius:6px;">8.3 A 2x2 grid</h3>

For more than 2 charts, you can create a grid with multiple rows and columns. In that case, <code>axes</code> becomes a 2D array, so you index it with <code>axes[row, col]</code>.


````python
fig, axes = plt.subplots(2, 2, figsize=(11, 8))

axes[0, 0].plot(days, nablus_temp, color="#38BDF8")
axes[0, 0].set_title("Line: Temperature over Days")

axes[0, 1].scatter(study_hours, exam_score, color="#BE185D")
axes[0, 1].set_title("Scatter: Study Hours vs. Score")

axes[1, 0].bar(tracks, num_interns, color="#FACC15")
axes[1, 0].set_title("Bar: Interns per Track")
axes[1, 0].tick_params(axis="x", rotation=20)

axes[1, 1].hist(ages, bins=15, color="#4ADE80")
axes[1, 1].set_title("Histogram: Age Distribution")

plt.tight_layout()
plt.show()
````

<h2 style="color:#BE185D; font-weight:bold; margin-top:30px;">10. Common Mistakes to Avoid</h2>

<div style="border-left:5px solid #BE185D; background-color:#fff0f6; padding:10px 15px; margin:10px 0; border-radius:4px; color:#7a0d3d;">
<b>Important:</b>
<ul>
<li><b>Forgetting axis labels or a title</b> — a chart with unlabeled axes forces the reader to guess what the numbers mean. Always label.</li>
<li><b>Wrong chart type for the data</b> — a line plot for unordered categories, or a bar chart for a continuous distribution, sends a misleading message.</li>
<li><b>Calling <code>plt.legend()</code> without ever setting <code>label=</code>d</b> — the legend box will appear, but empty or with default names like "line1".</li>
<li><b>Mixing up <code>plt.xxx()</code> and <code>axes[i].set_xxx()</code></b> inside subplots — remember the <code>set_</code> prefix once you're working with individual axes.</li>
<li><b>Not calling <code>plt.tight_layout()</code></b> in a subplot grid — titles and labels can overlap and become unreadable.</li>
<li><b>Reusing a "dirty" figure</b> — if you forget to start a new <code>plt.figure()</code>, a new plot can sometimes draw on top of an old one. When in doubt, start each chart with <code>plt.figure()</code>.</li>
</ul>

<h3 style="color:#FACC15; font-weight:bold; background-color:#1f2933; display:inline-block; padding:4px 10px; border-radius:6px;">Step 1 — Load and clean a dataset with Pandas</h3>
<div style="border-left:5px solid #38BDF8; background-color:#eaf6ff; padding:10px 15px; margin:10px 0; border-radius:4px; color:#0c4a6e;">
<b>Note:</b> We build a small sample "interns" dataset by hand (with a few missing values on purpose), exactly like a messy real-world CSV, and then clean it the way Day 4 taught us.
</div>

````python
np.random.seed(42)

data = {
    "name": ["Sara", "Omar", "Layla", "Ahmad", "Nour", "Yousef", "Rana",
              "Khaled", "Dima", "Fadi", "Huda", "Tamer"],
    "track": ["AI & ML", "Web Dev", "AI & ML", "Cybersecurity", "Data Analysis",
               "AI & ML", "Web Dev", "AI & ML", "Data Analysis", "Cybersecurity",
               "AI & ML", "Web Dev"],
    "age": [22, 24, np.nan, 21, 23, 25, 22, np.nan, 24, 26, 21, 23],
    "weekly_hours": [40, 38, 40, 35, 40, np.nan, 39, 40, 37, 40, 38, 40],
}

df = pd.DataFrame(data)

print("shape:", df.shape)
df.head()
````

````python
print("missing values per column:")
print(df.isnull().sum())

# fill missing numeric values with the column mean -- a simple, documented cleaning choice
df["age"] = df["age"].fillna(df["age"].mean())
df["weekly_hours"] = df["weekly_hours"].fillna(df["weekly_hours"].mean())

print("\nmissing values after cleaning:")
print(df.isnull().sum())
````

<h3 style="color:#FACC15; font-weight:bold; background-color:#1f2933; display:inline-block; padding:4px 10px; border-radius:6px;">Step 2 — Use NumPy to compute a derived feature</h3>
<div style="border-left:5px solid #38BDF8; background-color:#eaf6ff; padding:10px 15px; margin:10px 0; border-radius:4px; color:#0c4a6e;">
<b>Note:</b> We compute a z-score for each intern's age (how many standard deviations they are from the average) using NumPy's vectorized math — no loop required.
</div>

````python
ages_array = df["age"].to_numpy()          # convert the Pandas column into a NumPy array

mean_age = np.mean(ages_array)
std_age = np.std(ages_array)

df["age_zscore"] = (ages_array - mean_age) / std_age   # vectorized computation, applied to every row at once

print(f"mean age : {mean_age:.2f}")
print(f"std  age : {std_age:.2f}")
df[["name", "age", "age_zscore"]]
````

<h3 style="color:#FACC15; font-weight:bold; background-color:#1f2933; display:inline-block; padding:4px 10px; border-radius:6px;">Step 3 — Produce at least three labeled plots</h3>

<div style="border-left:5px solid #4ADE80; background-color:#eafff2; padding:10px 15px; margin:10px 0; border-radius:4px; color:#065f46;">
<b>Plot 1 — Histogram:</b> the distribution of intern ages, to see the overall age spread of the cohort.
</div>

````python
plt.figure(figsize=(7, 4))
plt.hist(df["age"], bins=8, color="#38BDF8", edgecolor="white")
plt.xlabel("Age")
plt.ylabel("Number of Interns")
plt.title("Distribution of Intern Ages")
plt.grid(True, alpha=0.3)
plt.show()
````

<div style="border-left:5px solid #4ADE80; background-color:#eafff2; padding:10px 15px; margin:10px 0; border-radius:4px; color:#065f46;">
<b>Plot 2 — Bar chart:</b> the number of interns per track, computed with a Pandas <code>groupby</code> (Day 4), to see which tracks are most popular.
</div>

````python
interns_per_track = df.groupby("track")["name"].count().sort_values(ascending=False)

plt.figure(figsize=(7, 4))
plt.bar(interns_per_track.index, interns_per_track.values, color="#FACC15")
plt.xlabel("Track")
plt.ylabel("Number of Interns")
plt.title("Interns per Track")
plt.xticks(rotation=15)
plt.show()
````

<div style="border-left:5px solid #4ADE80; background-color:#eafff2; padding:10px 15px; margin:10px 0; border-radius:4px; color:#065f46;">
<b>Plot 3 — Scatter plot:</b> age vs. weekly hours, to check whether older or younger interns tend to log more hours.
</div>

````python
plt.figure(figsize=(7, 4))
plt.scatter(df["age"], df["weekly_hours"], s=90, c="#BE185D", alpha=0.8, edgecolors="black")
plt.xlabel("Age")
plt.ylabel("Weekly Hours")
plt.title("Age vs. Weekly Hours")
plt.grid(True, alpha=0.3)
plt.show()
````

<div style="border-left:5px solid #4ADE80; background-color:#eafff2; padding:10px 15px; margin:10px 0; border-radius:4px; color:#065f46;">
<b>Bonus Plot — All three combined in one figure using subplots</b>, the way a final report would present them together.
</div>

````python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

axes[0].hist(df["age"], bins=8, color="#38BDF8", edgecolor="white")
axes[0].set_title("Age Distribution")
axes[0].set_xlabel("Age")
axes[0].set_ylabel("Count")

axes[1].bar(interns_per_track.index, interns_per_track.values, color="#FACC15")
axes[1].set_title("Interns per Track")
axes[1].set_xlabel("Track")
axes[1].set_ylabel("Count")
axes[1].tick_params(axis="x", rotation=25)

axes[2].scatter(df["age"], df["weekly_hours"], c="#BE185D", alpha=0.8, edgecolors="black")
axes[2].set_title("Age vs. Weekly Hours")
axes[2].set_xlabel("Age")
axes[2].set_ylabel("Weekly Hours")

plt.tight_layout()
plt.show()
````
