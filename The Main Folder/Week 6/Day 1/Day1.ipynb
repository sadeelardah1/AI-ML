Classical ML (Weeks 3–4), which this project's five-model comparison already used, works well on structured, tabular data — exactly the kind of data in <code>heart.csv</code>. <h2 style="color:#F2C14E; background-color:#2E6B4A; font-weight:bold; margin-top:30px; padding:8px 14px; border-radius:6px;">4. The Neuron</h2>
A single neuron is the building block, and it is exactly the operation from Week 2: it takes inputs, multiplies each by a weight, sums them, adds a bias, then passes the result through an activation function. In one line: <code>output = activation( dot(inputs, weights) + bias )</code>. A neural network is just many of these neurons connected in layers.

```python
# One neuron, conceptually, using two of this project's own scaled features:
# z = (Oldpeak * w1) + (MaxHR * w2) + bias      <- the dot product from Week 2
# output = activation(z)

import numpy as np

# Example values, standardized the same way the project's preprocessing pipeline does
oldpeak_scaled = 0.8   # a patient with an above-average Oldpeak
maxhr_scaled = -1.1    # a patient with a below-average max heart rate

x = np.array([oldpeak_scaled, maxhr_scaled])
w = np.array([0.6, -0.5])   # made-up starting weights
bias = 0.05

z = np.dot(x, w) + bias
output = 1 / (1 + np.exp(-z))   # sigmoid activation -> a probability

print('Weighted sum + bias (z):', round(z, 4))
print('Neuron output (probability of HeartDisease):', round(output, 4))
```

<div style="border-left:5px solid #4A9DE0; background-color:#eaf4fc; padding:10px 15px; margin:10px 0; border-radius:4px; color:#1c4e6e;">
<b>Note:</b> Notice both features chosen here — <code>Oldpeak</code> and <code>MaxHR</code> — are exactly the two numeric features this project's own EDA (from the classical ML phase) found to have the strongest correlation with <code>HeartDisease</code> (+0.404 and -0.401). A real neuron does not know this in advance; it starts with random weights and learns which features matter through training — but it is a useful sanity check that the raw math above already leans the right direction.
</div>

<h2 style="color:#F2C14E; background-color:#2E6B4A; font-weight:bold; margin-top:30px; padding:8px 14px; border-radius:6px;">5. Layers and Architecture</h2>
Neurons are organized into layers, and data flows through them in order:

<div style="border:1px solid #ccc; border-radius:6px; overflow:hidden;">
<table style="width:100%; border-collapse:collapse;">
<tr style="background-color:#2E6B4A; color:#F2C14E;"><th style="padding:8px; text-align:left;">Layer</th><th style="padding:8px; text-align:left;">Role</th><th style="padding:8px; text-align:left;">For This Project</th></tr>
<tr style="background-color:#ffffff;"><td style="padding:8px; border-top:1px solid #eee; color:#1a1a1a;">Input layer</td><td style="padding:8px; border-top:1px solid #eee; color:#1a1a1a;">Receives the raw features (one node per feature)</td><td style="padding:8px; border-top:1px solid #eee; color:#1a1a1a;">15 nodes — the same 15 numeric + one-hot encoded columns the classical models already use</td></tr><tr style="background-color:#ffffff;"><td style="padding:8px; border-top:1px solid #eee; color:#1a1a1a;">Hidden layer(s)</td><td style="padding:8px; border-top:1px solid #eee; color:#1a1a1a;">Learn intermediate representations; more/wider layers = more capacity</td><td style="padding:8px; border-top:1px solid #eee; color:#1a1a1a;">To be designed on Day 4</td></tr><tr style="background-color:#ffffff;"><td style="padding:8px; border-top:1px solid #eee; color:#1a1a1a;">Output layer</td><td style="padding:8px; border-top:1px solid #eee; color:#1a1a1a;">Produces the final prediction (shape depends on the task)</td><td style="padding:8px; border-top:1px solid #eee; color:#1a1a1a;">1 node, sigmoid activation — binary HeartDisease probability</td></tr>
</table>
</div>

"Deep" learning simply means a network with multiple hidden layers. Each layer transforms its input and passes the result forward, and stacking layers lets the network learn increasingly abstract patterns.

<a id="section6"></a>
<h2 style="color:#F2C14E; background-color:#2E6B4A; font-weight:bold; margin-top:30px; padding:8px 14px; border-radius:6px;">6. Weights and Biases Are What's Learned</h2>
The weights and biases are the network's parameters — the numbers adjusted during training. A fresh network starts with random weights and makes random predictions; training is the process of nudging these weights until the predictions become accurate. Everything on Days 2–3 this week is about how that nudging works.

<div style="border-left:5px solid #4A9DE0; background-color:#eaf4fc; padding:10px 15px; margin:10px 0; border-radius:4px; color:#1c4e6e;">
<b>Note:</b> The neuron is not new math — it is the Week 2 dot product plus a bias plus an activation. A neural network is that same operation, stacked and repeated. Keeping this in mind demystifies deep learning.
</div>

<a id="section7"></a>
<h2 style="color:#F2C14E; background-color:#2E6B4A; font-weight:bold; margin-top:30px; padding:8px 14px; border-radius:6px;">7. Common Mistakes to Avoid</h2>
<div style="border-left:5px solid #F0964B; background-color:#fef2e7; padding:10px 15px; margin:10px 0; border-radius:4px; color:#8a4a13;">
<b>Important:</b> <ul><li><b>Retraining the baseline from scratch instead of reusing it</b> — if a trustworthy baseline already exists (as it does here), formally confirming and recording it is enough; there is no need to duplicate work.</li><li><b>Assuming a neural network will automatically beat classical models</b> — on small, tabular datasets like this one (917 rows), classical models are often very competitive, sometimes better.</li><li><b>Forgetting the preprocessing pipeline already built</b> — the same <code>ColumnTransformer</code> logic in <code>src/preprocessing.py</code> should feed the neural network too, for a fair comparison.</li><li><b>Skipping formal Sprint Planning because 'the project is basically done'</b> — the Sprint discipline (goal, backlog, feature branches, PRs) still matters for documenting the deep-learning phase properly.</li><li><b>Comparing the neural network to the wrong baseline metric</b> — F1-score and ROC-AUC, not accuracy alone, are this project's agreed primary metrics because the classes are only mildly imbalanced (44.71% / 55.29%).</li></ul>
</div>
