<h2 style="color:#FFFFFF;background:linear-gradient(90deg,#0F766E,#0284C7);font-weight:bold;margin-top:30px;padding:10px 15px;border-radius:8px;">4. Build requirements.txt</h2>

<div style="background:#F8F8F8;color:#111;border:1px solid #E9E9E9;border-radius:9px;padding:16px 18px;line-height:1.7;">
<p>The hosting platform builds a fresh environment from <code>requirements.txt</code>. The next cell records exact versions from the environment that already runs the app.</p>
</div>

```python
from importlib.metadata import version, PackageNotFoundError
from pathlib import Path

packages = [
    ("streamlit", "streamlit"),
    ("scikit-learn", "scikit-learn"),
    ("joblib", "joblib"),
    ("numpy", "numpy"),
    ("nltk", "nltk"),
]

lines = []

for requirement_name, package_name in packages:
    try:
        v = version(package_name)
        line = f"{requirement_name}=={v}"
        lines.append(line)
        print(line)
    except PackageNotFoundError:
        print(f"[MISSING PACKAGE] {package_name}")

Path("requirements.txt").write_text(
    "\n".join(lines) + "\n",
    encoding="utf-8"
)

print("\nrequirements.txt created.")
```
